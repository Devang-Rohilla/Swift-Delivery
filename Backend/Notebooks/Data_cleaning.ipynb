#### removing ill-logical columns 

```python
data = data.drop(columns=['ID','Delivery_person_ID','Delivery_person_Age','Delivery_person_Ratings','Order_Date','Time_Orderd','Time_Order_picked'], errors='ignore')
```

```python
from Distance_calculation import haversine_distance
```

```python
coords = [
    "Restaurant_latitude",
    "Restaurant_longitude",
    "Delivery_location_latitude",
    "Delivery_location_longitude",
]
for col in coords:
    data[col] = data[col].abs()

data["distance_km"] = haversine_distance(
    data["Restaurant_latitude"],
    data["Restaurant_longitude"],
    data["Delivery_location_latitude"],
    data["Delivery_location_longitude"],
)
data["distance_km"]=data["distance_km"].round(2)
```
